```python
# 1º CÓDIGO: Geração do Banco de Dados CSV (Google Colab)

import numpy as np
import pandas as pd

# 1. Configurar semente aleatória para reprodutibilidade
np.random.seed(42)

# 2. Simular dados de profundidade das rugas após 30 dias (em milímetros)
dados_creme = np.random.normal(loc=1.85, scale=0.25, size=30)
dados_placebo = np.random.normal(loc=2.15, scale=0.30, size=30)

# 3. Criar DataFrame estruturado
df_estudo = pd.DataFrame({
    'ID_Voluntaria': range(1, 61),
    'Grupo': ['Creme Antirrugas'] * 30 + ['Placebo'] * 30,
    'Profundidade_Rugas_mm': np.concatenate([dados_creme, dados_placebo])
})

# 4. Exportar para arquivo CSV
df_estudo.to_csv('dados_estudo_cosmeticos.csv', index=False)
print("Arquivo 'dados_estudo_cosmeticos.csv' gerado com sucesso!")
```

```python
import numpy as np
import scipy.stats as stats
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Simulação dos Dados
np.random.seed(42)

# Grupo Placebo: 30 voluntárias, média de profundidade 2.5mm, desvio padrão 0.5
placebo = np.random.normal(loc=2.5, scale=0.5, size=30)

# Grupo Creme: 30 voluntárias, média de profundidade 2.1mm, desvio padrão 0.4
creme = np.random.normal(loc=2.1, scale=0.4, size=30)

df = pd.DataFrame({
    'Profundidade_Rugas': np.concatenate([placebo, creme]),
    'Grupo': ['Placebo']*30 + ['Creme']*30
})

# 2. Verificação dos Pressupostos
# Teste de Normalidade (Shapiro-Wilk)
_, p_shapiro_placebo = stats.shapiro(placebo)
_, p_shapiro_creme = stats.shapiro(creme)

# Teste de Homogeneidade das Variâncias (Levene)
_, p_levene = stats.levene(placebo, creme)

# 3. Teste t de Student para Amostras Independentes
# Hipótese unilateral (alternative='less'): testamos se a média do Creme é MENOR que a do Placebo.
t_stat, p_valor = stats.ttest_ind(
    a=creme,
    b=placebo,
    equal_var=(p_levene >= 0.05),
    alternative='less'
)

# 4. Exibição dos Resultados
print("--- RESULTADOS DOS PRESSUPOSTOS ---")
print(f"Normalidade Placebo (Shapiro-Wilk): p = {p_shapiro_placebo:.4f}")
print(f"Normalidade Creme (Shapiro-Wilk): p = {p_shapiro_creme:.4f}")
print(f"Homocedasticidade (Levene): p = {p_levene:.4f}")
print("\n--- RESULTADO DO TESTE T DE STUDENT ---")
print(f"Estatística t: {t_stat:.4f}")
print(f"Valor P: {p_valor:.4f}")
print("-" * 40)

if p_valor < 0.05:
    print("Conclusão: Rejeitamos a Hipótese Nula (H0).\nO creme antirrugas apresentou eficácia estatisticamente significativa na redução da profundidade das rugas.")
else:
    print("Conclusão: Falhamos em rejeitar a Hipótese Nula (H0).\nNão há evidências estatísticas suficientes para afirmar que o creme reduz a profundidade das rugas.")

# 5. Visualização Gráfica
plt.figure(figsize=(10, 6))
sns.kdeplot(data=df, x='Profundidade_Rugas', hue='Grupo', fill=True, palette={'Placebo': '#e74c3c', 'Creme': '#3498db'}, alpha=0.6)

plt.axvline(np.mean(placebo), color='#c0392b', linestyle='dashed', linewidth=2, label=f'Média Placebo: {np.mean(placebo):.2f}mm')
plt.axvline(np.mean(creme), color='#2980b9', linestyle='dashed', linewidth=2, label=f'Média Creme: {np.mean(creme):.2f}mm')

plt.title('Distribuição da Profundidade das Rugas após 30 Dias de Teste', fontsize=14)
plt.xlabel('Profundidade das Rugas (mm)', fontsize=12)
plt.ylabel('Densidade de Probabilidade', fontsize=12)
plt.legend()
plt.tight_layout()
plt.show()
```
