```python
! wget -O breaking-bad-s01e01.txt https://gist.githubusercontent.com/alvelvis/e08b9a85796ca4bbfb6a71cfadf3b291/raw/74efcee1d9fa98dc13893ed67b1e6d9cef075548/breaking-bad-subtitles-pt-s01ep01

# Abrindo a legenda pra entendendo sua estrutura
with open("breaking-bad-s01e01.txt") as f:
    legenda = f.read()

# Transformando a legenda em um texto só
falas = legenda.split("\n\n")
texto = ""
for fala in falas:
    partes = fala.split("\n", 2)
    if len(partes) > 2:
        texto = texto + " " + partes[2]

# Limpa as legendas
texto = texto.replace("\n", " ")
texto = texto.replace("<i>", "")
texto = texto.replace("</i>", "")
texto = texto.replace("<b>", "")
texto = texto.replace("</b>", "")

# Segmenta as frases
frases = texto.split(". ")
frases = [frase + "." for frase in frases]

# Segmenta os tokens
frases = [frase\
          .replace(".", " . ")\
          .replace(",", " , ")\
          .replace("!", " ! ")\
          .replace("?", " ? ")\
          .replace(":", " : ")\
          .replace(" ", " ")\
          .strip() for frase in frases]

frases_tokenizadas = [frase.split(" ") for frase in frases]

"""
Substituição do yake e spacy por estruturas nativas (dicionários, listas, tuplas),
pois a atividade exige que não seja utilizado bibliotecas externas.
"""
texto_normalizado = texto.lower()

# 1. Limpeza manual de pontuações
pontuacoes = [".", ",", "!", "?", ":", ";", "(", ")", "-", '"', "'", "♪"]
for p in pontuacoes:
    texto_normalizado = texto_normalizado.replace(p, f" {p} ")

tokens_gerais = texto_normalizado.split()

# 2. Tupla de Stopwords (Estrutura imutável nativa)
stop_words = (
    "o", "a", "os", "as", "um", "uma", "uns", "umas",
    "e", "ou", "mas", "se", "como", "porque",
    "de", "do", "da", "dos", "das",
    "em", "no", "na", "nos", "nas",
    "para", "com", "por", "sem",
    "não", "sim", "que", "é", "são", "foi", "era",
    "eu", "você", "ele", "ela", "nós", "eles", "elas",
    "me", "te", "se", "lhe", "nos", "vos", "lhes",
    "meu", "minha", "seu", "sua", "nosso", "nossa",
    "este", "esta", "esse", "essa", "isso", "aquilo",
    "aqui", "ali", "lá", "aí", "onde",
    "ser", "estar", "ter", "fazer", "ir", "quero",
    "já", "só", "ao", "aos", "muito", "mais", "bem",
    "qual", "quais", "quem", "tudo", "está", "tá", "vai", "vou"
)

# 3. Filtragem de tokens (Remoção de stopwords, pontuações e números)
tokens_limpos = []
for token in tokens_gerais:
    if token not in stop_words and token not in pontuacoes and not token.isdigit():
        tokens_limpos.append(token)

# 4. Extração de Palavras-Chave por Frequência (usando Dicionário nativo)
frequencia_keywords = {}
for palavra in tokens_limpos:
    frequencia_keywords[palavra] = frequencia_keywords.get(palavra, 0) + 1

# 5. Ordenação decrescente (Convertendo dicionário para Lista de Tuplas)
keywords_nativas = sorted(list(frequencia_keywords.items()), key=lambda x: x[1], reverse=True)

# 6. Exibição simulando a saída das bibliotecas
print("--- Palavras-Chave Mais Frequentes (Top 20) ---")
for keyword, frequencia in keywords_nativas[:20]:
    print(f"{keyword}: {frequencia}")

print("\n--- Entidades/Tokens Únicos (Simulando o Set do spaCy) ---")
# Usando set nativo para mostrar tokens únicos como o spaCy faria com entidades
entidades_simuladas = set(tokens_limpos[:50]) # Limitado aos 50 primeiros para demonstração
for token in entidades_simuladas:
    print(token)
```

```python
def processar_texto_simulado(texto):
    """
    Simula as principais etapas de pré-processamento de texto:
    tokenização, remoção de stopwords, lematização e contagem de frequência.
    """

    # 1. Tokenização e normalização
    # Em uma aplicação real, essa etapa poderia ser realizada
    # utilizando recursos de bibliotecas de Processamento de Linguagem Natural.
    texto_normalizado = texto.lower()

    pontuacoes = [".", ",", "!", "?", ":", ";", "(", ")"]

    for p in pontuacoes:
        texto_normalizado = texto_normalizado.replace(p, f" {p} ")

    tokens = texto_normalizado.split()

    # 2. Remoção de Stopwords e Lematização
    # Em uma aplicação real, a lematização poderia ser realizada
    # com ferramentas como spaCy, reduzindo as palavras à sua forma base.
    stop_words = (
        "o", "a", "os", "as",
        "um", "uma",
        "e", "ou", "mas",
        "de", "do", "da",
        "em", "para", "com",
        "não", "é", "que"
    )

    tokens_limpos = []

    for token in tokens:
        if token not in stop_words and token not in pontuacoes:
            tokens_limpos.append(token)

    # 3. Contagem de frequência
    # Em uma aplicação real, essa etapa poderia utilizar estruturas
    # ou recursos específicos de bibliotecas de análise textual.
    frequencia = {}

    for palavra in tokens_limpos:
        frequencia[palavra] = frequencia.get(palavra, 0) + 1

    # Ordenação da frequência das palavras em ordem decrescente.
    frequencia_lista = list(frequencia.items())

    frequencia_ordenada = sorted(
        frequencia_lista,
        key=lambda x: x[1],
        reverse=True
    )

    return {
        "tokens_originais": tokens,
        "tokens_processados": tokens_limpos,
        "frequencia_ordenada": frequencia_ordenada
    }
```

```python
# Exemplo de uso da função processar_texto_simulado
texto_exemplo = "O processamento de linguagem natural é uma subárea da inteligência artificial. O PLN estuda a interação entre computadores e linguagens humanas."

# Chamada da função
resultado = processar_texto_simulado(texto_exemplo)

# Exibição dos resultados estruturados
print("--- Tokens Originais ---")
print(resultado['tokens_originais'])

print("\n--- Tokens Processados (Sem Stopwords) ---")
print(resultado['tokens_processados'])

print("\n--- Frequência de Palavras (Ordenada) ---")
for palavra, qtd in resultado['frequencia_ordenada']:
    print(f"{palavra}: {qtd}")
```
