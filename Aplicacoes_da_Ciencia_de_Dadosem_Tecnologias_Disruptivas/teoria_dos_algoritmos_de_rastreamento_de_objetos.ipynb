```python
from google.colab import files

arquivos = files.upload()
video_entrada = next(iter(arquivos))
```

```python
import cv2
from ultralytics import YOLO
from IPython.display import Video, display

modelo = YOLO("yolo11n.pt")

captura = cv2.VideoCapture(video_entrada)

largura = int(captura.get(cv2.CAP_PROP_FRAME_WIDTH))
altura = int(captura.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = captura.get(cv2.CAP_PROP_FPS)

if fps <= 0:
    fps = 30

video_saida = "rastreamento_pessoas.mp4"

codec = cv2.VideoWriter_fourcc(*"mp4v")
gravador = cv2.VideoWriter(
    video_saida,
    codec,
    fps,
    (largura, altura)
)

while captura.isOpened():
    sucesso, frame = captura.read()

    if not sucesso:
        break

    resultados = modelo.track(
        frame,
        persist=True,
        tracker="bytetrack.yaml",
        classes=[0],
        conf=0.35,
        iou=0.5,
        verbose=False
    )

    frame_processado = resultados[0].plot()

    quantidade = 0

    if resultados[0].boxes is not None:
        if resultados[0].boxes.id is not None:
            quantidade = len(resultados[0].boxes.id)

    cv2.putText(
        frame_processado,
        f"Pessoas rastreadas: {quantidade}",
        (20, 40),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (0, 255, 0),
        2
    )

    gravador.write(frame_processado)

captura.release()
gravador.release()

print("Rastreamento concluído.")
```
